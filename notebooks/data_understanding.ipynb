# 02 Data Cleaning

## Aim

The aim of this notebook is to create an analysis-ready dataset from the raw meningitis dataset while preserving the original raw data unchanged.

## Raw Data Principle

The original dataset in `data/raw/meningitis.csv` will not be modified. All cleaning decisions will be applied to a copied dataframe and saved as a new file in `data/processed/`.

## Column Selection Decision

The raw dataset contains baseline variables, follow-up variables, treatment-related variables, second lumbar puncture variables, and diagnostic confirmation variables.

For the first analysis, I kept variables that are available at baseline or are clinically relevant for early differentiation between bacterial and viral meningitis.

Variables excluded from the first cleaned dataset include unclear variables, treatment/follow-up variables, second lumbar puncture variables, and diagnostic confirmation variables that may leak the final diagnosis.

## Variables Kept

The following variables were kept:

- `year`
- `month`
- `age`
- `race`
- `sex`
- `priorrx`
- `wbc`
- `pmn`
- `bands`
- `bloodgl`
- `gl`
- `pr`
- `reds`
- `whites`
- `polys`
- `lymphs`
- `monos`
- `others`
- `gram`
- `subset`
- `abm`

## Variables Excluded For The First Analysis

The following variables were excluded from the first cleaned dataset:

- `dx`
- `priordx`
- `compns`
- `daysrx`
- `offrx`
- `lptodc`
- `lpgap`
- `morelabs`
- `culture`
- `cie`
- `bloodclt`
- `bloodgl2`
- `gl2`
- `pr2`
- `reds2`
- `whites2`
- `polys2`
- `lymphs2`
- `monos2`
- `others2`
- `sumbands`

## Rationale

These variables were excluded because they are unclear, treatment-related, follow-up variables, second lumbar puncture measurements, or diagnostic confirmation variables. Since the goal of this project is to evaluate early baseline predictors of bacterial meningitis, including post-baseline or final diagnostic confirmation variables could bias the analysis.

## Column Renaming

Variables were renamed to improve readability while preserving a record of the original names in the data dictionary.

```python
# columns to keep 

columns_to_keep = ["year", "month", "age", "race", "sex", "priorrx",
    "wbc", "pmn", "bands", "bloodgl", "gl", "pr",
    "reds", "whites", "polys", "lymphs", "monos", "others",
    "gram", "subset", "abm"]

# renaming map 
rename_map ={
    "year": "case_year",
    "month": "case_month",
    "age": "age_years",
    "priorrx": "prior_antibiotics",
    "wbc": "blood_wbc_count",
    "pmn": "blood_pmn_percent",
    "bands": "blood_band_percent",
    "bloodgl": "blood_glucose_mg_dl",
    "gl": "csf_glucose_mg_dl",
    "pr": "csf_protein_mg_dl",
    "reds": "csf_rbc_count",
    "whites": "csf_wbc_count",
    "polys": "csf_pmn_percent",
    "lymphs": "csf_lymphocyte_count",
    "monos": "csf_monocyte_value",
    "others": "csf_other_cell_value",
    "gram": "gram_stain_result",
    "subset": "original_subset",
    "abm": "bacterial_meningitis"
}

# conversion 
df_clean = df[columns_to_keep].copy()

df_clean = df_clean.rename(columns=rename_map)
display(df_clean.head())
```

```python
# export the dataset 

df_clean.to_csv('../data/processed/meningitis_clean_v1.csv')
```
